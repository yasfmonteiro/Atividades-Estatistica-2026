````python
import pandas as pd
import matplotlib.pyplot as plt
````

````python
df = pd.read_excel("dados_semana02.xlsx")

df.head()
````

````python
contagem = df["Produto"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(contagem,
        labels=contagem.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Distribuição dos Produtos")
plt.legend(contagem.index, title="Produtos")
plt.show()
````

````python
media_preco = df.groupby("Produto")["Preco"].mean()

plt.figure(figsize=(6,4))
plt.bar(media_preco.index,
        media_preco.values)

plt.title("Preço Médio por Produto")
plt.xlabel("Produto")
plt.ylabel("Preço Médio")
plt.show()
````

````python
plt.figure(figsize=(6,4))

plt.boxplot(df["Vendas"])

plt.title("Distribuição das Vendas")
plt.ylabel("Vendas")
plt.show()
````

````python
plt.figure(figsize=(6,4))

plt.hist(df["Preco"],
         bins=6)

plt.title("Distribuição dos Preços")
plt.xlabel("Preço")
plt.ylabel("Frequência")
plt.show()
````
