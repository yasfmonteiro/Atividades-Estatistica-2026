## 1. Distribuição de probabilidade e parâmetros
- Variável aleatória ($X$): Peso líquido de uma embalagem de farinha selecionada aleatoriamente, em gramas ($\text{g}$).   
- Distribuição: $X \sim N(\mu = 1005, \sigma^2 = 8^2)$   
- Parâmetros: Média ($\mu$): $1005\text{ g}$   Desvio-padrão ($\sigma$): $8\text{ g}$   

##2. Probabilidade de peso inferior ou igual a 1.000 g
- Notação: $P(X \le 1000)$   
- Cálculo de padronização ($Z$):$$Z = \frac{1000 - 1005}{8} = -0,625$$
- Resultado:$$P(X \le 1000) \approx 0,2660 \text{ ou } 26,60\%$$

## 3. Gráfico da probabilidade $P(X \le 1000)$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros
mu = 1005
sigma = 8

# Eixo X para o gráfico
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, loc=mu, scale=sigma)

# Probabilidade
p_1000 = norm.cdf(1000, loc=mu, scale=sigma)

# Plot
plt.figure(figsize=(8, 4.5))
plt.plot(x, y, label=r'$X \sim N(1005, 8^2)$', color='blue', lw=2)

# Sombreado P(X <= 1000)
x_fill = np.linspace(mu - 4*sigma, 1000, 500)
y_fill = norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color='red', alpha=0.4, label=f'P(X ≤ 1000) = {p_1000:.4f}')

plt.axvline(1000, color='red', linestyle='--', label='X = 1000 g')
plt.title('Distribuição Normal: Probabilidade de Peso ≤ 1000 g')
plt.xlabel('Peso (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

## 4. Probabilidade de peso superior ou igual a 1.015 g
- Notação: $P(X \ge 1015)$   
- Cálculo de padronização ($Z$):$$Z = \frac{1015 - 1005}{8} = 1,25$$
- Resultado:$$P(X \ge 1015) = 1 - P(X < 1015) = 1 - 0,8944 = 0,1057 \text{ ou } 10,57\%$$

## 5. Gráfico da probabilidade $P(X \ge 1015)$

```python
# Probabilidade P(X >= 1015)
p_1015 = 1 - norm.cdf(1015, loc=mu, scale=sigma)

# Plot
plt.figure(figsize=(8, 4.5))
plt.plot(x, y, label=r'$X \sim N(1005, 8^2)$', color='blue', lw=2)

# Sombreado P(X >= 1015)
x_fill = np.linspace(1015, mu + 4*sigma, 500)
y_fill = norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color='green', alpha=0.4, label=f'P(X ≥ 1015) = {p_1015:.4f}')

plt.axvline(1015, color='green', linestyle='--', label='X = 1015 g')
plt.title('Distribuição Normal: Probabilidade de Peso ≥ 1015 g')
plt.xlabel('Peso (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

## 6. Determinação do valor $x$ tal que $P(X \le x) = 0,90$
- Notação: Percentil 90 da distribuição.   
- Cálculo:$$Z = \Phi^{-1}(0,90) \approx 1,2816$$$$x = \mu + Z \cdot \sigma = 1005 + (1,2816 \cdot 8) \approx 1015,25\text{ g}$$

```python
# Percentil 90
x_p90 = norm.ppf(0.90, loc=mu, scale=sigma)
print(f"O valor de x tal que P(X <= x) = 0.90 é: {x_p90:.2f} g")

# Plot do Percentil 90
plt.figure(figsize=(8, 4.5))
plt.plot(x, y, label=r'$X \sim N(1005, 8^2)$', color='blue', lw=2)

x_fill = np.linspace(mu - 4*sigma, x_p90, 500)
y_fill = norm.pdf(x_fill, loc=mu, scale=sigma)
plt.fill_between(x_fill, y_fill, color='orange', alpha=0.4, label=f'Área de 90% (x = {x_p90:.2f} g)')

plt.axvline(x_p90, color='orange', linestyle='--', label=f'P90 = {x_p90:.2f} g')
plt.title('Percentil 90 da Distribuição de Pesos')
plt.xlabel('Peso (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

## 7. Interpretação no contexto da linha de produção
O valor de $1015,25\text{ g}$ indica que 90% das embalagens produzidas nessa linha apresentam peso de até $1015,25\text{ g}$. Em contrapartida, apenas 10% das embalagens produzidas ultrapassam esse limite de peso, o que auxilia no monitoramento de excesso de envase (overfill) e no controle de custos de matéria-prima.
