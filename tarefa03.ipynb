```python
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    "Fruta_preferida": [
        "Banana", "Morango", "Banana", "Maçã", "Morango",
        "Banana", "Laranja", "Maçã", "Banana", "Morango",
        "Laranja", "Banana", "Maçã", "Morango", "Banana",
        "Laranja", "Maçã", "Banana", "Morango", "Banana"
    ],

    "Porcoes_frutas_semana": [
        7, 5, 10, 4, 6,
        8, 3, 5, 9, 6,
        4, 12, 7, 5, 10,
        3, 6, 8, 4, 9
    ],

    "Idade": [
        18, 19, 18, 20, 19,
        21, 18, 20, 19, 22,
        18, 21, 20, 19, 18,
        22, 20, 19, 21, 18
    ]
}

df = pd.DataFrame(dados)

df
```

```python
freq_fruta = df["Fruta_preferida"].value_counts()

freq_relativa_fruta = df["Fruta_preferida"].value_counts(normalize=True) * 100

tabela_fruta = pd.DataFrame({
    "Frequência absoluta": freq_fruta,
    "Frequência relativa (%)": freq_relativa_fruta
})

tabela_fruta
```

A banana foi a fruta preferida pela maior parcela dos estudantes, representando 40% das respostas. O morango apareceu em segundo lugar, com 25%, seguido pela maçã (20%) e pela laranja (15%).

```python
freq_porcoes = df["Porcoes_frutas_semana"].value_counts().sort_index()

freq_relativa_porcoes = (
    df["Porcoes_frutas_semana"]
    .value_counts(normalize=True)
    .sort_index() * 100
)

tabela_porcoes = pd.DataFrame({
    "Frequência absoluta": freq_porcoes,
    "Frequência relativa (%)": freq_relativa_porcoes
})

tabela_porcoes
```

Observa-se uma variação no número de porções de frutas consumidas semanalmente pelos estudantes. Os valores variaram de 3 a 12 porções por semana, indicando diferentes hábitos de consumo dentro do grupo analisado.

```python
plt.figure(figsize=(7,5))

freq_fruta.plot(kind="bar", color='lightpink')

plt.title("Fruta Preferida pelos Estudantes")
plt.xlabel("Fruta")
plt.ylabel("Número de estudantes")
plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.hist(df["Porcoes_frutas_semana"], bins=14, color='lightpink', edgecolor='black')

plt.title("Distribuição do Consumo Semanal de Frutas")
plt.xlabel("Porções de frutas por semana")
plt.ylabel("Número de estudantes")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(
    df["Idade"],
    df["Porcoes_frutas_semana"],
    color='lightpink'
)

plt.title("Relação entre Idade e Consumo de Frutas")
plt.xlabel("Idade (anos)")
plt.ylabel("Porções de frutas por semana")

plt.show()
```

O gráfico de dispersão não indica uma relação linear forte entre idade e número de porções de frutas consumidas por semana. Os valores estão relativamente dispersos, sugerindo que, nessa amostra, a idade não parece ser um fator determinante para o consumo de frutas.
